```python
from abc import ABC, abstractmethod

class Fiugura(ABC): # Base (Padre)
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimetro(self):
        pass

    def mostrar(self):
        print("Es un figura geometrica")
```

```python
class Cuadrado(Fiugura): # Derivada (Hija)
    def __init__(self, lado):
        self.lado = lado

    def area(self):
        return self.lado ** 2

    def perimetro(self):
        return 4 * self.lado

    def info(self):
        print(f"Cuadrado de lado {self.lado}")

    def metodo_propio(self):
        print("Metodo propio de la clase Cuadrado")
```

```python
objeto1 = Cuadrado(5)
objeto1.info()
print("Area:", objeto1.area())
print("Perimetro:", objeto1.perimetro())
```

```python
figura1 = Fiugura()  # Esto generará un error porque Fiugura es una clase abstracta y no se puede instanciar directamente
figura1.mostrar()
```

# Ejercicio 1

```python
class Accionar():
    def encender(self):
        print("Encendiendo...")
    def apagar(self):
        print("Apagando...")
```

```python
from abc import ABC, abstractmethod

class Electrodomestico(ABC):
    def __init__(self,potenciaKw,marca):
        self.potenciaKw = potenciaKw
        self.marca = marca

    @abstractmethod
    def ConsumoEnergetico(self,horas):
        pass

    def __str__(self):
        return f"Electrodomestico de marca {self.marca} con potencia {self.potenciaKw} kW"

    def encender(self,objetoAccionar):
        if isinstance(objetoAccionar, Accionar):
            objetoAccionar.encender()
        else:
            print("El objeto no es de tipo Accionar")

    def apagar(self,objetoAccionar):
        if isinstance(objetoAccionar, Accionar):
            objetoAccionar.apagar()
        else:
            print("El objeto no es de tipo Accionar")
```

```python
class Coccion(Electrodomestico):
    def __init__(self,potenciaKw,marca,programable):
        super().__init__(potenciaKw,marca)
        self.programable = programable

    def ConsumoEnergetico(self,horas):
        return self.potenciaKw * horas

    def __str__(self):
        return f"Coccion de marca {self.marca} con potencia {self.potenciaKw} kW y programable: {self.programable}"
```

```python
class Cocina(Coccion):
    def __init__(self,potenciaKw,marca,programable,hornillas):
        super().__init__(potenciaKw,marca,programable)
        self.hornillas = hornillas

    def ConsumoEnergetico(self,horas):
        return super().ConsumoEnergetico(horas) * self.hornillas

    def __str__(self):
        return f"Cocina de marca {self.marca} con potencia {self.potenciaKw} kW y {self.hornillas} hornillas"
```

```python
class Horno(Coccion):
    def __init__(self,potenciaKw,marca,programable,capacidad):
        super().__init__(potenciaKw,marca,programable)
        self.capacidad = capacidad

    def ConsumoEnergetico(self,horas):
        return super().ConsumoEnergetico(horas) * self.capacidad

    def __str__(self):
        return f"Horno de marca {self.marca} con potencia {self.potenciaKw} kW y capacidad {self.capacidad} litros"
```

```python
# Craer obejto accionar
objetoAccionar = Coccion(4,4,4)

# crear objeto cocina y horno
cocina1 = Cocina(2.5, "Samsung", True, 4)
horno1 = Horno(3.0, "LG", True, 50)

#mostrar informacion de los objetos
print(cocina1)
cocina1.encender(objetoAccionar)
cocina1.apagar(objetoAccionar)
print("Consumo energetico de la cocina en 2 horas:", cocina1.ConsumoEnergetico(2), "kWh")

print(horno1)
horno1.encender(objetoAccionar)
horno1.apagar(objetoAccionar)
print("Consumo energetico del horno en 3 horas:", horno1.ConsumoEnergetico(3), "kWh")
```

```python
from abc import ABC, abstractmethod

class Volumen(ABC):
    def __init__(self, id, name):
        self.id = id
        self.name = name

    def __str__(self):
        pass

    @property
    def id(self):
        return self._id
    @id.setter
    def id(self, value):
        self._id = value

    @property
    def name(self):
        return self._name
    @name.setter
    def name(self, value):
        self._name = value
```

```python
class Book(Volumen):
    def __init__(self, id, name, isbn_code):
        super().__init__(id, name)
        self.isbn_code = isbn_code

    def __str__(self):
        return f"Book ID: {self.id}, Name: {self.name}, ISBN: {self.isbn_code}"
```

```python
class Newpaper(Volumen):
    def __init__(self, id, name, issn_code):
        super().__init__(id, name)
        self.issn_code = issn_code

    def __str__(self):
        return f"Newpaper ID: {self.id}, Name: {self.name}, ISSN: {self.issn_code}"
```

```python
import pandas as pd

class Library:
    def __init__(self,name,address):
        self.name = name
        self.address = address
        self.volumes = []

    def add_volume(self, volume):
        for volumen in self.volumes:
            if volumen.id == volume.id:
                print(f"Volumen con ID {volume.id} ya existe.")
                return
        self.volumes.append(volume)

    def remove_volume(self, id_eliminar):
        for volumen in self.volumes:
            if volumen.id == id_eliminar:
                self.volumes.remove(volumen)
                return f"Volumen con ID {id_eliminar} eliminado."
            
        return f"No se encontró un volumen con ID {id_eliminar}."

    def save_volumes(self):
        data = []
        for volumen in self.volumes:
            if isinstance(volumen, Book):
                data.append([volumen.id, volumen.name, "Book", volumen.isbn_code])
            elif isinstance(volumen, Newpaper):
                data.append([volumen.id, volumen.name, "Newspaper", volumen.issn_code])
        
        df = pd.DataFrame(data, columns=["ID", "Name", "Type", "Code"])

        df.to_excel("semana11_taller.xlsx", index=False)
        print("Volúmenes guardados en 'semana11_taller.xlsx'.")

    def list_volumes(self):
        for volumen in self.volumes:
            print(volumen)     
```

```python
# Crear biblioteca
biblioteca = Library("Mi Biblioteca", "Calle Falsa 123")

# Crear libros y periódicos
libro1 = Book(1, "El Quijote", "978-3-16-148410-0")
libro2 = Book(2, "Cien Años de Soledad", "978-0-06-088328-7")
periodico1 = Newpaper(3, "El País", "1234-5678")
periodico2 = Newpaper(4, "La Nación", "8765-4321")

#Registrar libros y periódicos en la biblioteca
biblioteca.add_volume(libro1)
biblioteca.add_volume(libro2)
biblioteca.add_volume(periodico1)
biblioteca.add_volume(periodico2)


# Listar volúmenes
print("Listado de volúmenes en la biblioteca:")
biblioteca.list_volumes()

# Eliminar un volumen
id_eliminar = 2

#Después de eliminar, listar volúmenes
print("Volumen eliminado:")
print(biblioteca.remove_volume(id_eliminar))
print("Listado de volúmenes en la biblioteca:")
biblioteca.list_volumes()

# Guardar volúmenes en un archivo Excel
biblioteca.save_volumes()
```
